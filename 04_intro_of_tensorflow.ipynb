We can create random tensor by using `tf.random.Generator` class

```python
#create two random (but same) tensors
random_1 = tf.random.Generator.from_seed(42)
random_1 = random_1.normal(shape=(3,2)) #create tensor from a normal distribution
random_2 = tf.random.Generator.from_seed(42)
random_2 = random_2.normal(shape=(3,2)) #create tensor from a normal distribution
```

```python
#are they equal
random_1, random_2, random_1 == random_2
```

#Shuffle the order of element in tensor

```python
not_shuffled = tf.constant([[4,9],
                            [2,3],
                            [7,1]])

tf.random.shuffle(not_shuffled)
```

 If both the global and the operation seed are set: Both seeds are used in conjunction to determine the random sequence."

tf.random.set_seed(42) sets the global seed, and the seed parameter in tf.random.shuffle(seed=42) sets the operation seed.

Because, "Operations that rely on a random seed actually derive it from two seeds: the global and operation-level seeds. This sets the global seed.

```python
tf.random.set_seed(42) #Global level random seed
tf.random.shuffle(not_shuffled, seed=(42)) #Operation level random seed
```

#Getting information from tensor

* Shape
* Rank
* Axis or dimension
* Size

```python
#Create a rank 4 tensor (4 dimensions)
rank_4_tensor = tf.zeros(shape=(2,3,4,5))
rank_4_tensor
```

```python
rank_4_tensor.shape, rank_4_tensor.ndim, tf.size(rank_4_tensor)
```

```python
#Get various attributes of our tensor
print("Data type of every element:", rank_4_tensor.dtype)
print("Number of dimension:", rank_4_tensor.ndim)
print("Shape of tensor:", rank_4_tensor.shape)
print("Element along the 0 axis:", rank_4_tensor.shape[0])
print("Element along the last axis:", rank_4_tensor.shape[-1])
print("Total no. of elements:", tf.size(rank_4_tensor))
print("Total no. of elements:", tf.size(rank_4_tensor).numpy())
```

#Indexing tensor

Tensor cna be indexed just like Python list.

```python
#Get the first element form each dimension form each index exvept for the final one
rank_4_tensor[:1, :1, :1, :]
```

```python
#Rank 2 tensor
rank_2_tensor = tf.constant([[3,7],
                             [4,9]])
rank_2_tensor.shape, rank_2_tensor.ndim
```

```python
#Get the last item of each of our rank 2 tensor
rank_2_tensor[:, 1:] #or [:, -1]
```

```python
#Add extra dimension to our rank 2 tensor
rank_3_tensor = rank_2_tensor[..., tf.newaxis]
rank_3_tensor
```

```python
#Alternative to tf.newaxis
tf.expand_dims(rank_2_tensor, axis=-1) #-1 means expand the final axis
```

```python
#Expand the 0 axis
tf.expand_dims(rank_2_tensor, axis=0)
```

#Matrix multiplication

```python
#Matrix multiplication with tensorflow
tf.matmul(tensor, tensor)
```

```python
#With python operator @
tensor @ tensor
```

```python
#create a (3,2) tensor
X = tf.constant([[1,2],
                 [3,4],
                 [5,6]])
#another
Y = tf.constant([[7,8],
                [9,10],
                [11,12]])
X.shape, tf.reshape(Y, shape=(2,3)).shape
```

```python
#We have to change the shape of Y (or X)
tf.matmul(X, tf.reshape(Y, shape=(2,3)))
```

```python
#Can do same with transpose
X, tf.transpose(X)
```

```python
tf.matmul(tf.transpose(X), Y)
```

```python
tf.matmul(tf.reshape(X, (2,3)), Y)
```

in matrix multiplication always use transpose.

#Aggregating tensor

condensing them  from multiple values down to a smaller amount of values

```python
#Get the absolute value
x = tf.constant([-7, 10])
tf.abs(x)
```

**let's go through the following forms of aggrigation**

* get minimum
* get maximum
* get the mean of a tensor
* get the sum of a tensor

```python
# create a random tensor of value between 0 to 100 of size 50
r = tf.constant(np.random.randint(0, 100, size=50))
r
```

```python
tf.size(r), r.shape, r.ndim
```

```python
#find the minimun
tf.reduce_min(r)
```

```python
#find the maximum
tf.reduce_max(r)
```

```python
#Mean
tf.reduce_mean(r)
```

```python
#Sum
tf.reduce_sum(r)
```

```python
#to find the variance of our tensor we need access to tensorflow_probability
import tensorflow_probability as tfp
tfp.stats.variance(r)
```

```python
tf.math.reduce_variance(tf.cast(r, dtype=tf.float32))  #we have to change datatype to float32
```

```python
tf.math.reduce_std(tf.cast(r, dtype=tf.float32))
```

#Find the positional maximum and minimum

```python
#New tensor
tf.random.set_seed(42)
F = tf.random.uniform(shape=[50])
F
```

```python
#find teh positional maximum
tf.argmax(F)
```

```python
#index on our largest value position
F[tf.argmax(F)]
```

```python
#Find the max value of F
tf.reduce_max(F)
```

```python
#check for equality
F[tf.argmax(F)] == tf.reduce_max(F)
```
